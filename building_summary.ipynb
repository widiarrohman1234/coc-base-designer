```python
from pathlib import Path
import pandas as pd

from core.static_data_loader import StaticDataLoader
from core.export_parser import ExportParser
from core.building_summary import BuildingSummary
```

```python
# Root project
ROOT = Path("C:/xampp/htdocs/coc-base-designer")

# Load data
loader = StaticDataLoader(ROOT / "data" / "static_data.json")
print("\n=== Static Data ===")
print("Total building type :", loader.total_buildings())


parser = ExportParser(ROOT / "data" / "data_export_th8.json")
export_data = parser.load()

print("=== Export JSON ===")
print(export_data.keys())

print("\nJumlah building :", len(export_data["buildings"]))
print("Jumlah traps    :", len(export_data["traps"]))
print("Jumlah deco     :", len(export_data["decos"]))


summary = BuildingSummary(export_data, loader)
result = summary.summarize()
print("\n=== Village Summary ===")

print("Town Hall Level :", result["townhall_level"])
print("Building Count  :", result["building_count"])
print("Wall Count      :", result["wall_count"])
print("Map Size        :", result["map_size"])
```
