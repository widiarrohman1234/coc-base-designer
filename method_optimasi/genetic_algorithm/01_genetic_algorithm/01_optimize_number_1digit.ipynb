# GA satu digit

mencari nilai f(x)=0
$$
f(x)=−(x−5)^2
$$

```python
import random
```

```python
# =========================================
# 1. Parameter Genetic Algorithm
# =========================================

POPULATION_SIZE = 5
CHROMOSOME_LENGTH = 4
GENERATIONS = 15

MUTATION_RATE = 0.1
```

```python
# =========================================
# 2. Fungsi Fitness
# =========================================

def fitness(x):
    return -(x - 5) ** 2


# =========================================
# 3. Membuat chromosome random
# =========================================

def create_chromosome():
    return [
        random.randint(0, 1)
        for _ in range(CHROMOSOME_LENGTH)
    ]


# =========================================
# 4. Konversi binary → integer
# =========================================

def decode(chromosome):
    binary = ''.join(map(str, chromosome))
    return int(binary, 2)


# =========================================
# 5. Membuat population
# =========================================

def create_population():
    return [
        create_chromosome()
        for _ in range(POPULATION_SIZE)
    ]


# =========================================
# 6. Selection
# =========================================

def selection(population):
    population_sorted = sorted(
        population,
        key=lambda chromosome: fitness(decode(chromosome)),
        reverse=True
    )

    # Ambil 50% terbaik
    return population_sorted[:POPULATION_SIZE // 2]


# =========================================
# 7. Crossover
# =========================================

def crossover(parent1, parent2):

    point = random.randint(
        1,
        CHROMOSOME_LENGTH - 1
    )

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2


# =========================================
# 8. Mutation
# =========================================

def mutation(chromosome):

    for i in range(CHROMOSOME_LENGTH):

        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]

    return chromosome


```

```python
# =========================================
# 9. Genetic Algorithm
# =========================================

population = create_population()


for generation in range(GENERATIONS):

    # -----------------------------
    # Evaluasi
    # -----------------------------

    best = max(
        population,
        key=lambda chromosome: fitness(
            decode(chromosome)
        )
    )

    best_x = decode(best)
    best_fitness = fitness(best_x)

    print(
        f"Generation {generation + 1:02d} | "
        f"x = {best_x:2d} | "
        f"fitness = {best_fitness:3d}"
    )


    # -----------------------------
    # Selection
    # -----------------------------

    parents = selection(population)


    # -----------------------------
    # Membuat population baru
    # -----------------------------

    new_population = parents.copy()


    while len(new_population) < POPULATION_SIZE:

        parent1, parent2 = random.sample(parents, 2)

        child1, child2 = crossover(
            parent1,
            parent2
        )

        child1 = mutation(child1)
        child2 = mutation(child2)

        new_population.append(child1)

        if len(new_population) < POPULATION_SIZE:
            new_population.append(child2)


    population = new_population


# =========================================
# 10. Hasil akhir
# =========================================

best = max(
    population,
    key=lambda chromosome: fitness(
        decode(chromosome)
    )
)

best_x = decode(best)

print("\nHasil akhir")
print("Chromosome :", best)
print("x          :", best_x)
print("fitness    :", fitness(best_x))
```

Bagian-bagian GA 1 digit

| Konsep GA  | Dalam program        |
| ---------- | -------------------- |
| Gene       | 1 bit (`0` atau `1`) |
| Chromosome | `[1,0,1,0]`          |
| Individual | Satu chromosome      |
| Population | Kumpulan chromosome  |
| Decode     | Binary → integer     |
| Fitness    | `-(x-5)^2`           |
| Selection  | Memilih 50% terbaik  |
| Crossover  | Menggabungkan parent |
| Mutation   | Membalik bit         |
| Generation | Satu siklus evolusi  |
