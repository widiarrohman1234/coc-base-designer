# GA empat digit

mencari nilai f(x,y)=0
$$
f(x,y,z,a)=−(x−3)^2−(y−5)^2-(z-9)^2-(a-1)^2+100
$$

kunci jawabannya adalah x=3,y=5, z=9 dam a=1. f(3,5,9,1)=100

Namun dalam Genetic Algorithm, kita anggap kita tidak tahu bahwa (3,5,9,1) adalah jawabannya.

```python
import random
```

```python
# =========================================
# 1. Parameter Genetic Algorithm
# =========================================

POPULATION_SIZE = 20
GENERATIONS = 15

MIN_VALUE = 0
MAX_VALUE = 10

MUTATION_RATE = 0.1
```

```python
# =========================================
# 2. Fungsi Fitness
# =========================================

def fitness(individual):
    x, y, z, a = individual
    return -(x - 3) ** 2 - (y - 5) ** 2 - (z - 9) ** 2 - (a - 1) ** 2 + 100


# =========================================
# 3. Membuat individual
# =========================================

def create_individual():

    x = random.randint(MIN_VALUE, MAX_VALUE)
    y = random.randint(MIN_VALUE, MAX_VALUE)
    z = random.randint(MIN_VALUE, MAX_VALUE)
    a = random.randint(MIN_VALUE, MAX_VALUE)

    return [x, y, z, a]


# =========================================
# 5. Membuat population
# =========================================

def create_population():
    return [
        create_individual()
        for _ in range(POPULATION_SIZE)
    ]


# =========================================
# 6. Selection
# =========================================

def selection(population):
    population_sorted = sorted(
        population,
        key=fitness,
        reverse=True
    )

    # Ambil 50% terbaik
    return population_sorted[:POPULATION_SIZE // 2]


# =========================================
# 7. Crossover
# =========================================

def crossover(parent1, parent2):

    child1 = [
        parent1[0],
        parent2[1],
        parent1[2],
        parent2[3]
    ]

    child2 = [
        parent2[0],
        parent1[1],
        parent2[2],
        parent1[3],
    ]

    return child1, child2


# =========================================
# 8. Mutation
# =========================================

def mutation(individual):
    for i in range(4):
        if random.random() < MUTATION_RATE:
            individual[i] = random.randint(
                MIN_VALUE,
                MAX_VALUE
            )

    return individual


```

```python
# =========================================
# 9. Genetic Algorithm
# =========================================

population = create_population()


for generation in range(GENERATIONS):

    # -----------------------------
    # Evaluasi
    # -----------------------------

    best = max(
        population,
        key=fitness
    )

    print(
        f"Generation {generation + 1:02d} | "
        f"x = {best[0]:2d} | "
        f"y = {best[1]:2d} | "
        f"z = {best[2]:2d} | "
        f"a = {best[3]:2d} | "
        f"fitness = {fitness(best):.2f}"
    )


    # -----------------------------
    # Selection
    # -----------------------------

    parents = selection(population)


    # -----------------------------
    # Membuat population baru
    # -----------------------------

    new_population = parents.copy()


    while len(new_population) < POPULATION_SIZE:

        parent1, parent2 = random.sample(parents, 2)

        child1, child2 = crossover(
            parent1,
            parent2
        )

        child1 = mutation(child1)
        child2 = mutation(child2)

        new_population.append(child1)

        if len(new_population) < POPULATION_SIZE:
            new_population.append(child2)


    population = new_population


# =========================================
# 10. Hasil akhir
# =========================================

best = max(
    population,
    key=fitness
)

print("\nHasil akhir")
print("x       :", best[0])
print("y       :", best[1])
print("z       :", best[2])
print("a       :", best[3])
print("fitness :", fitness(best))
```

Bagian-bagian GA 2 digit

| Konsep GA      | Dalam program                                                                      |
| -------------- | ---------------------------------------------------------------------------------- |
| **Gene**       | Satu nilai/variabel, yaitu `x`, `y`, `z` dan `a`                                            |
| **Chromosome** | `[x, y, z, a]`, misalnya `[3, 5, 9, 1]`                                                        |
| **Individual** | Satu chromosome                                                                    |
| **Population** | Kumpulan chromosome                                                                |
| **Fitness**    | Nilai dari fungsi `f(x,y) = -(x-3)²-(y-5)²-(z-9)²-(a-1)²+100`                                    |
| **Selection**  | Memilih 50% individual dengan fitness terbaik                                      |
| **Crossover**  | 4 genre                                     |
| **Mutation**   | Mengubah nilai gene `x`,`y`,`z` dam `a` secara random                                     |
| **Generation** | Satu siklus evolusi: evaluasi → selection → crossover → mutation → population baru |
