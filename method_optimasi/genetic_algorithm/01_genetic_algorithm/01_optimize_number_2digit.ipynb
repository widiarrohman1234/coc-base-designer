# GA dua digit

mencari nilai f(x,y)=0
$$
f(x,y)=−(x−3)^2−(y−5)^2+100
$$

kunci jawabannya adalah x=3 dan y=5. f(3,5)=100

Namun dalam Genetic Algorithm, kita anggap kita tidak tahu bahwa (3,5) adalah jawabannya.

```python
import random
```

```python
# =========================================
# 1. Parameter Genetic Algorithm
# =========================================

POPULATION_SIZE = 10
GENERATIONS = 15

X_MIN = 0
X_MAX = 10

Y_MIN = 0
Y_MAX = 10

MUTATION_RATE = 0.1
```

```python
# =========================================
# 2. Fungsi Fitness
# =========================================

def fitness(individual):
    x, y = individual
    return -(x - 3) ** 2 - (y - 5) ** 2 + 100


# =========================================
# 3. Membuat individual
# =========================================

def create_individual():

    x = random.randint(X_MIN, X_MAX)
    y = random.randint(Y_MIN, Y_MAX)

    return [x, y]


# =========================================
# 5. Membuat population
# =========================================

def create_population():
    return [
        create_individual()
        for _ in range(POPULATION_SIZE)
    ]


# =========================================
# 6. Selection
# =========================================

def selection(population):
    population_sorted = sorted(
        population,
        key=fitness,
        reverse=True
    )

    # Ambil 50% terbaik
    return population_sorted[:POPULATION_SIZE // 2]


# =========================================
# 7. Crossover
# =========================================

def crossover(parent1, parent2):

    child1 = [
        parent1[0],
        parent2[1]
    ]

    child2 = [
        parent2[0],
        parent1[1]
    ]

    return child1, child2


# =========================================
# 8. Mutation
# =========================================

def mutation(individual):

    x, y = individual

    if random.random() < MUTATION_RATE:
        x = random.randint(X_MIN, X_MAX)

    if random.random() < MUTATION_RATE:
        y = random.randint(Y_MIN, Y_MAX)

    return [x, y]


```

```python
# =========================================
# 9. Genetic Algorithm
# =========================================

population = create_population()


for generation in range(GENERATIONS):

    # -----------------------------
    # Evaluasi
    # -----------------------------

    best = max(
        population,
        key=fitness
    )

    print(
        f"Generation {generation + 1:02d} | "
        f"x = {best[0]:2d} | "
        f"y = {best[1]:2d} | "
        f"fitness = {fitness(best):.2f}"
    )


    # -----------------------------
    # Selection
    # -----------------------------

    parents = selection(population)


    # -----------------------------
    # Membuat population baru
    # -----------------------------

    new_population = parents.copy()


    while len(new_population) < POPULATION_SIZE:

        parent1, parent2 = random.sample(parents, 2)

        child1, child2 = crossover(
            parent1,
            parent2
        )

        child1 = mutation(child1)
        child2 = mutation(child2)

        new_population.append(child1)

        if len(new_population) < POPULATION_SIZE:
            new_population.append(child2)


    population = new_population


# =========================================
# 10. Hasil akhir
# =========================================

best = max(
    population,
    key=fitness
)

print("\nHasil akhir")
print("x       :", best[0])
print("y       :", best[1])
print("fitness :", fitness(best))
```

Bagian-bagian GA 2 digit

| Konsep GA      | Dalam program                                                                      |
| -------------- | ---------------------------------------------------------------------------------- |
| **Gene**       | Satu nilai/variabel, yaitu `x` atau `y`                                            |
| **Chromosome** | `[x, y]`, misalnya `[3, 5]`                                                        |
| **Individual** | Satu chromosome                                                                    |
| **Population** | Kumpulan chromosome                                                                |
| **Fitness**    | Nilai dari fungsi `f(x,y) = -(x-3)²-(y-5)²+100`                                    |
| **Selection**  | Memilih 50% individual dengan fitness terbaik                                      |
| **Crossover**  | Menggabungkan gene `x` dan `y` dari dua parent                                     |
| **Mutation**   | Mengubah nilai gene `x` atau `y` secara random                                     |
| **Generation** | Satu siklus evolusi: evaluasi → selection → crossover → mutation → population baru |
